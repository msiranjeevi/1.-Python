```python
#Extraction
#string -> "hello" -> sequence of characters
#           each character has a position called as index. Address of each character
#           we have postive indexes(+) and negative indexes(-)
#           positive indexes -> "h e l l o"
#            from left to right  0 1 2 3 4 
#           negative indexes -> "h  e  l  l  o"
#            from right to left -5 -4 -3 -2 -1 
#           In order to use index use [] and mention the index value in it
#       Extract multiple characters is called as Slicing
#       Syntax: [start:end] -> start is included and end is not included
#       Ex: [0:3] -> hel     [-5:-2] -> hel    [1:] -> ello    [1] -> e
#       Syntax: [start:end:step] -> step says how many characters to be skipped  
#       Ex: [0:4:2] -> hl   -> step 1 is default
                        
                        
                        
text="Python"

#Extract First
print(text[0])
print(text[-6])

#Extract Last
print(text[5])
print(text[-1])

#Extract h
print(text[3])
print(text[-3])

#Slicing
date='2026-09-03'

#Extract year
print(date[0:4])

#open-ended slicing
print(date[:4])  #default open index is 0
print(date[-10:-6])

#Extract month
print(date[5:7])
print(date[-5:-3])

#Extract date
print(date[8:])
print(date[-2:])                        
                
```

String - Cleaning

```python
#Cleansing
#1.1 Clean whitespaces
#1.2 Clean cases

#1.1 Clean whitespaces
# Remove whitespaces
#What is whitespace? "    Max"    
#                       |----> this is the whitespace left side
#To clear the left justified spaces, use a method lstrip()
#                  "Max   "    
#                       |----> this is the whitespace right side
#To clear the right justified spaces, use a method rstrip()
#To clear both, use a method strip()

########Clean it before you perform a search#########################
#Method 1
text1="     banking"
print(text1.lstrip())

text2="banking     "  
print(text2.rstrip())
  
text3="      banking       " 
print(text3.strip())

print("-" * 50)
#Method 2
text_a="     Wellsfargo".lstrip()
print(text_a)

text_b="Wellsfargo        ".rstrip()
print(text_b)

text_c="      Wellsfargo       ".strip()
print(text_c)

print("-" * 50)
#Strip other than whitespaces

sam_text="###Jacob###".strip("#")
print(sam_text)

print("-" * 50)
#How do you check whether whitespaces occured in start or end? 
#Detect extra spaces

text_bank="     Wellsfargo"
print(len(text_bank)) 
print(len(text_bank.strip()))

print("-" * 50)

print(len(text_bank) - len(text_bank.strip()))
print(len(text_bank) == len(text_bank.strip()))

print("-" * 50)

no_of_spaces = len(text_bank) - len(text_bank.strip())
is_data_clean = len(text_bank) == len(text_bank.strip())
print("No. of spaces", no_of_spaces)
print("Is the data clean?", is_data_clean)

print("*" * 50)

#1.1 Clean cases - case conversion
# 2 methods -> upper() and lower()

text_user="python PROGRAMMING"

#clean data for matching
print(text_user.upper())

print(text_user.lower())

print("-" * 50)

#case not converted
search="Email"
mail_data="email"

print(search == mail_data)

print("-" * 50)

#case converted
search="Email".lower()
mail_data="email".lower()

print(search == mail_data)

print("-" * 50)

#case converted with strip for extra space
search="Email ".lower().strip()
mail_data="email".lower()

print(search == mail_data)

```
