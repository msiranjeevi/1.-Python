```python
#prime number check
def pnc(no):
    if no > 1:
        for i in range(2,no):
            if no % i == 0:
                print("Its not a prime")
                break
        else:
            print("Its a prime")    
    #else:
    #    print("Its not a prime")            
```

```python
pnc(8)
```

```python
def fac(fno):
    c=1
    for i in range(1,fno+1):
        c = c * i
    print(c)        
```

```python
def arm(no):
    arm_list=list(str(no))
    int_arm_list=[int(x) for x in arm_list]
    print(arm_list)
    print(int_arm_list)
    c = 0
    for i in int_arm_list:
        c = c + (i ** 3)
        if c == no: 
            print("Its an armstrong number")
    else:
        print("Its not an armstrong number") 
```

```python
arm(123)
```
