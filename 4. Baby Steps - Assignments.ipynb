```python
ht = int(input("Enter your height(cms):"))
wt = int(input("Enter your weight(kgs)"))
ht_square = ht * ht
bmi = wt / (ht / 100) ** 2
r_bmi = round(bmi,1)
if r_bmi > 0 and r_bmi <= 18.5:
    print("You are under weight")
elif r_bmi >= 18.5 and r_bmi <= 24.9:
    print("You are on idle weight")
elif r_bmi >= 25 and r_bmi <= 29.9:
    print("You are over weight")
elif r_bmi >=30 and r_bmi <= 34.9:
    print("You are in Obseity Class-I")
elif r_bmi >= 35 and r_bmi <= 39.9:
    print("You are in Obseity Class-II")
elif r_bmi >= 40:
    print("You are in Obseity Class-III, Severe Class")    
             
print(r_bmi)
```
