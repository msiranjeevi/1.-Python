```python

no1=float(input("Enter number 1:"))
no2=float(input("Enter number 2"))

#Addition
a_res = no1 + no2
print("Addition Result:", a_res)

#Subtraction
s_res = no1 - no2
print("Subtraction Result:", s_res)

#Multiplication
m_res = no1 * no2
print("Multiplication Result:", m_res)

#Division
d_res = no1 / no2
print("Division Result:,", d_res)

#Floor Division
d_res = no1 // no2
print("Floor Division Result:", d_res)

#Modulo
m_res = no1 % no2
print("Modulo Result is:", m_res)

#Power
p_res = no1 ** no2
print("Square value is:", p_res)

#input number investigation
if (no1 % 2) == 0:
    print("no1 is a even number")
else:
    print("no1 is not an even number")    
```
