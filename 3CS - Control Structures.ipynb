IF Statement

```python
#Age group check
age=int(input("Enter your age:"))
if age>=1 and age<=17:
    print("You are a minor, not qualified to cast a vote")
else:
    print("Hurray! You are qualified to cast a vote")
        
```

```python
#Age group check
age=int(input("Enter your age:"))
if age>=0 and age<=17:
    print("You are a minor, not qualified to cast a vote")
elif age>=18 and age<=22:
    print("You are a first generation voter") 
elif age>22 and age<=50:
    print("You are a second generation voter") 
elif age>50 and age<=100:
    print("You are a senior citizen, pick up and drop arranged from govt.")              
elif age<0:
    print("Age is invalid!")    
else:
    print("You can cast vote virtually!!!!")
        
```
