```python
#odd and even from a list
nos = [11, 20, 31, 42, 53, 80]
for i in nos:
    if (i % 2) == 0:
        print(i, "is an even number")
    else:
        print(i, "is an odd number")
        
    
```

```python
#odd and even numbers seperately to a list
nos = [11, 20, 31, 42, 53, 80]
odd_nos = []
eve_nos = []
for i in nos:
    if (i % 2) == 0:
        eve_nos.append(i)
        
    else:
        odd_nos.append(i)
print("Odd Numbers List:", odd_nos)
print("Even Numbers List:", eve_nos)
        
    
```
