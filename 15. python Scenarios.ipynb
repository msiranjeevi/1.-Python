3. 10% bonus for employees who are getting salary >= $50000

```python
salaries=[23000,54000,45000,62000]
ctr = 0
for i in salaries:
    ctr = ctr + 1
    if i >= 50000:
        print("Employee number:",ctr)
        print("Your salary is:", i)
        bonus = i
        bonus = (bonus * 10 / 100)
        print("Your 10% bonus is", bonus)
        print("Your total salary is:", round(i + bonus,2))
        print("*********************************")
        
    else:
        print("Employee number:",ctr)   
        print("Your salary is:", i)
        print("You are not eligible for bonus")
        print("********************************")
        
      
```

7. A retail store - WELCOME TO COSTCO LIMITED

```python
totalOrderValues=[123,99,134,156]
print("     WELCOME TO COSTCO LIMITED")
print("Today's offer : 20% off for $100 aand above")
print("~~~~~~~~~~~~~~~~~~~~~~~~~~")
for i in totalOrderValues:
    if i >= 100:
        print("Your total purchased value:", i)
        dis = (i * 20 / 100)
        print("You have 20% of discount of:",dis)
        print("Total Payable:", round((i - dis),2))
        print("---------------------------------------")
    else:
        print("Your total purchased value:", i)
        print("You dont have discounts bcoz values is less than $100")    
        print("---------------------------------------")
```

9. Leap Year or not

```python
#If a year is divisible by 400 and 4 are leap years
#If a year is divisible by 100 then its not
year = int(input("Enter a year:"))
if year % 400 == 0:
    print(year , "is a leap year")
elif year % 100 == 0:
    print(year, "is not a leap year")    
elif year % 4 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")           
    
```
